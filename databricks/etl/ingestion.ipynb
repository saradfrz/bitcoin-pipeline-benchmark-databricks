# Binance BTCUSDT Daily Data Ingestion

Description: Downloads and extracts the previous day’s Binance BTCUSDT trade data, using Brasília time to determine the date. Saves CSV files to the Databricks landing-zone volume for Bronze-layer ingestion and raises an error if no files are returned.

```python
from project_config import config
databricks_project_root = config["databricks_project_root"]

import sys
sys.path.insert(0,f"{config["databricks_project_root"]}/src")
from binance_downloader import BinanceTradeDownloader
```

```python

from datetime import datetime, timedelta
from zoneinfo import ZoneInfo

# Use Brasília time explicitly
now = datetime.now(ZoneInfo("America/Sao_Paulo"))
yesterday = (now.date() - timedelta(days=1)).strftime("%d/%m/%Y")

print(f"Current time: {now}")
print(f"Downloading: {yesterday}")

downloader = BinanceTradeDownloader(
    start_date=yesterday,
    end_date=yesterday,
    volume="/Volumes/btc_trading/binance_bronze/landing_zone",
    symbol="BTCUSDT",
    sleep_seconds=1.0
)

files = downloader.run()

if not files:
    raise RuntimeError(f"Binance archive not available for {yesterday}")

```
