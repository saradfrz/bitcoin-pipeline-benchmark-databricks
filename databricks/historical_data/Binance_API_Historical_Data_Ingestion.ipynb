# Binance BTCUSDT Historical Data Ingestion

Downloads and extracts Binance BTCUSDT trade data from 01/01/2025, saving CSV files to the Databricks landing-zone volume for subsequent Bronze-layer ingestion. Uses project configuration to load the downloader module and pauses one second between requests.


```python

from project_config import config
databricks_project_root = config["databricks_project_root"]

import sys
from pathlib import Path
sys.path.insert(0,f"{config["databricks_project_root"]}/src")

from binance_downloader import BinanceTradeDownloader
```

```python
volume = Path("/Volumes/btc_trading/binance_bronze/landing_zone")

downloader = BinanceTradeDownloader(
    start_date="30/10/2026",
    end_date="30/10/2026",
    volume=str(volume),
    symbol="BTCUSDT",
)

csv_files = downloader.run()
```
